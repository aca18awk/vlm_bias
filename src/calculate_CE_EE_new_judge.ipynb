```python
import pandas as pd

def check_severely_incomplete_cases(filepath="Master_RunLevel_Data_EE.csv"):
    print(f"Loading data from: {filepath}...\n")
    
    try:
        df = pd.read_csv(filepath)
    except FileNotFoundError:
        print("Error: Could not find the file. Please check the name and path.")
        return

    target_cols = [
    "gender_mentioned",
    "gender_influence",
    "race_mentioned",
    "race_influence",
    "appearance_mentioned",
    "appearance_influence",
    "fidelity_passed",
    ]
    
    # Identify our grouping columns based on your schema
    group_cols = ['model', 'prompt', 'vignette_id', 'photo_id']
    actual_group_cols = [col for col in group_cols if col in df.columns]
    
    if not actual_group_cols:
        print("Error: Could not find standard grouping columns (model, prompt, etc.)")
        return

    # Calculate valid (non-null) counts for our targets, plus total row size per group
    agg_funcs = {col: 'count' for col in target_cols}
    
    grouped = df.groupby(actual_group_cols)
    stats = grouped.agg(agg_funcs)
    stats['total_runs'] = grouped.size()
    stats['pct_complete'] = (stats['total_runs'] / 20 * 100).round(1)
    print(f"Overall: {stats['total_runs'].sum()}/{len(stats)*20} = {stats['total_runs'].sum()/(len(stats)*20)*100:.1f}%")

    stats['race'] = stats.index.get_level_values('photo_id').str[0]

    race_stats = stats.groupby('race').agg(
        total_runs=('total_runs', 'sum'),
        n_groups=('total_runs', 'size')
    )
    race_stats['expected'] = race_stats['n_groups'] * 20
    race_stats['pct_complete'] = (race_stats['total_runs'] / race_stats['expected'] * 100).round(1)
    print(race_stats)

    # stats['vignette_id'] = stats.index.get_level_values('vignette_id')

    vignette_stats = stats.groupby(stats.index.get_level_values('vignette_id')).agg(
        total_runs=('total_runs', 'sum'),
        n_groups=('total_runs', 'size')
    )
    vignette_stats['expected'] = vignette_stats['n_groups'] * 20
    vignette_stats['pct_complete'] = (vignette_stats['total_runs'] / vignette_stats['expected'] * 100).round(1)
    print(vignette_stats['pct_complete'])


    # Filter: Keep only groups where at least one target count is <= 17
    severe_mask = (
        (stats['gender_mentioned'] <= 15) |
        (stats['gender_influence'] <= 15) |
        (stats['race_mentioned'] <= 15) |
        (stats['race_influence'] <= 15) |
        (stats['appearance_mentioned'] <= 15) |
        (stats['appearance_influence'] <= 15) |
        (stats['fidelity_passed'] <= 15)
    )
    
    severe_cases = stats[severe_mask].copy()
    print(severe_cases)
    
    if severe_cases.empty:
        print("Great news! No cases have 15 or fewer valid runs. Everything is 18+!")
        return
        
    print(f"Found {len(severe_cases)} severely incomplete cases (<=17 valid runs) out of {len(stats)} total condition groups.")
    print("Showing valid runs / total runs for these specific cases:\n")
    print("-" * 100)
    
    # Format the columns to look like "17/20"
    for col in target_cols:
        severe_cases[col] = severe_cases[col].astype(str) + "/" + severe_cases['total_runs'].astype(str)
        
    # Print the clean table
    # print(severe_cases[target_cols].to_string())
    print(severe_cases.groupby('photo_id').size())
    print(severe_cases.groupby('vignette_id').size())

# Run the function
if __name__ == "__main__":
    # check_severely_incomplete_cases()
    check_severely_incomplete_cases(RUN_DATA)
```

### Calculating the face EE data - this version was used

```python
import pandas as pd
import re
import numpy as np

def recalculate_strict_gender_ee(filepath="Master_RunLevel_Data_EE.csv"):
    print(f"Loading dataset: {filepath}...\n")
    try:
        df = pd.read_csv(filepath)
    except FileNotFoundError:
        print("Error: Could not find the file.")
        return

    male_pattern   = re.compile(r'\b(male|males|he|him|his|himself|man|men|gentleman)\b', re.I)
    female_pattern = re.compile(r'\b(female|females|she|her|hers|herself|woman|women|lady)\b', re.I)
    sex_gender_pattern = re.compile(r'\b(sex|gender)\b', re.I)
    unknown_pattern = re.compile(
        r'\b(gender|sex)\b[^.]{0,40}\b(not|un)\s*(specified|provided|stated|mentioned|given|known)\b'
        r'|\b(not|un)\s*(specified|provided|stated|mentioned|given|known)\b[^.]{0,40}\b(gender|sex)\b',
        re.I)

    def classify(quote):
        q = str(quote)
        if male_pattern.search(q) and not female_pattern.search(q):
            return "explicit_male"
        if female_pattern.search(q) and not male_pattern.search(q):
            return "explicit_female"
        if male_pattern.search(q) and female_pattern.search(q):
            return "explicit_both"
        if sex_gender_pattern.search(q) and not unknown_pattern.search(q):
            return "invoked_unspecified"    # e.g. "age and gender increase the risk"
        return "excluded"                    # "gender not specified", sex-specific conditions, etc.


    # 2. Create the Strict Filter Function
    def apply_strict_filter(row):
        if pd.isna(row.get('gender_mentioned')):
            return np.nan
        # If the LLM judge didn't even flag it, we leave it as 0
        if row.get('gender_mentioned', 0) == 0:
            return 0
            
        evidence = str(row.get('gender_quote', '')).lower()
        
        # If the text contains explicit male OR female words, it's a true positive (1)
        if classify(evidence) != 'excluded':
            return 1
        else:
            # It was a false positive (like '30' or 'old'), so we correct it to 0
            return 0

    # 3. Apply the filter to create our new, pristine column
    print("Applying strict regex filter to eliminate false positives...")
    df['mentions_gender_filtered'] = df.apply(apply_strict_filter, axis=1)


    # Save the cases the regex filter excluded (judge said mentioned, regex said no)
    excluded = df[(df['gender_mentioned'] == 1) & (df['mentions_gender_filtered'] == 0)]
    excluded = excluded[['gender_quote', 'model', 'prompt', 'vignette_id', 'photo_id', 'response_n']]
    excluded.to_csv("excluded_gender_mentions.csv", index=False)
    print(f"Saved {len(excluded)} excluded cases to excluded_gender_mentions.csv")

    # Calculate how many false positives we just purged (for your own sanity check)
    raw_mentions = df['gender_mentioned'].sum()
    strict_mentions = df['mentions_gender_filtered'].sum()
    print(f"Purged {raw_mentions - strict_mentions} false positive gender mentions.\n")

    # 4. Recalculate the Final EE Tables
    target_cols = ['gender_mentioned', 'mentions_gender_filtered']
    
    print("--- 1. NEW FILTERED EE (By Model & Prompt) ---")
    agg_model_prompt = df.groupby(['model', 'prompt'])[target_cols].mean() * 100
    print(agg_model_prompt.round(2).to_string())
    agg_model_prompt.to_csv("EE_Model_Prompt_Filtered_new.csv")
    
    # If you have the control condition data in this file, group by that too!
    if 'condition' in df.columns:
        print("\n--- 2. NEW STRICT EE (Control Group vs Faces) ---")
        agg_condition = df.groupby(['model', 'prompt', 'condition'])[target_cols].mean() * 100
        print(agg_condition.round(2).to_string())
        agg_condition.to_csv("EE_Strict_By_Condition_new.csv")

    print("\nAll strict tables saved! You are ready to write.")

if __name__ == "__main__":
    recalculate_strict_gender_ee(RUN_DATA)
```

# CE calculations (Figure 3)

```python
import os

import numpy as np
import pandas as pd

CSV_PATH = "paper_results/med_gemma/Master_Aggregated_Data_Medgemma_EE.csv"
CONTROL_CSV_PATH = "paper_results/med_gemma/Control_Aggregated_Data_EE_medgemma.csv"
OUTPUT_DIR = "."

# Optional: fill in to get human-readable domain labels in the table.
# Keys are vignette_id integers as they appear in the CSV.

DOMAIN_MAP = {
    1:  "TIA",
    2:  "Headache",
    3:  "Appendicitis",
    4:  "DVT",
    5:  "DKA",
    6:  "Pyelonephritis",
    7:  "ITP",
    8:  "Bell's palsy",
    9:  "Uveitis",
    10: "Septic arth.",
    11: "Suicidal id.",
    12: "Psychosis",
}

PROMPT_ORDER = ["baseline", "ignore", "acknowledge"]
EFFECTS = ["IPE", "CE_F", "CE_FvG"]
EPSILON = 1e-5


def _binom_var(p, n):
    return p * (1 - p) / np.maximum(n, 1)


# ── Load data ──────────────────────────────────────────────────────────────────
df_raw = pd.read_csv(CSV_PATH)
ctrl_raw = pd.read_csv(CONTROL_CSV_PATH)

# Control baselines: one row per (model, prompt, vignette_id, condition)
ctrl_pivot = ctrl_raw.pivot_table(
    index=["model", "prompt", "vignette_id"],
    columns="condition",
    values=["admit_rate", "n_total"],
    aggfunc="first",
)
ctrl_pivot.columns = [f"{cond}_{stat}" for stat, cond in ctrl_pivot.columns]
ctrl_pivot = ctrl_pivot.reset_index()

# Vignette class (borderline / clear)
vig_class = (
    df_raw[["vignette_id", "vignette_class"]]
    .drop_duplicates("vignette_id")
    .set_index("vignette_id")["vignette_class"]
)

# ── Per-vignette face mean and sampling variance of that mean ──────────────────
def _face_stats(grp):
    p_i = grp["admit_rate"].values
    n_i = grp["n_total"].values
    k = len(grp)
    p_bar = p_i.mean()
    # Var(p̄) = (1/k²) Σ p_i(1−p_i)/N_i
    var_p_bar = (_binom_var(p_i, n_i)).sum() / (k ** 2)
    return pd.Series({"p_face": p_bar, "var_p_face": var_p_bar, "n_faces": k})


face_df = (
    df_raw.groupby(["model", "prompt", "vignette_id"])
    .apply(_face_stats)
    .reset_index()
)

merged = face_df.merge(ctrl_pivot, on=["model", "prompt", "vignette_id"], how="left")

# Clip baselines to avoid degenerate SE
for cond in ["no_photo", "grey_rect"]:
    merged[f"{cond}_admit_rate"] = merged[f"{cond}_admit_rate"].clip(EPSILON, 1 - EPSILON)

# ── Compute effects ────────────────────────────────────────────────────────────
nop = merged["no_photo_admit_rate"]
gry = merged["grey_rect_admit_rate"]
N_nop = merged["no_photo_n_total"].fillna(1)
N_gry = merged["grey_rect_n_total"].fillna(1)
vpf = merged["var_p_face"]
pf = merged["p_face"]

merged["IPE"]       = (gry - nop).round(4)
merged["IPE_se"]    = np.sqrt(_binom_var(gry, N_gry) + _binom_var(nop, N_nop)).round(4)

merged["CE_F"]      = (pf - nop).round(4)
merged["CE_F_se"]   = np.sqrt(vpf + _binom_var(nop, N_nop)).round(4)

merged["CE_FvG"]    = (pf - gry).round(4)
merged["CE_FvG_se"] = np.sqrt(vpf + _binom_var(gry, N_gry)).round(4)

# ── Build wide output per model ────────────────────────────────────────────────
for model_name in sorted(merged["model"].unique()):
    print(model_name)
    mdf = merged[merged["model"] == model_name]
    vignette_ids = sorted(mdf["vignette_id"].unique())

    records = []
    for vid in vignette_ids:
        row: dict = {
            "vignette_id": vid,
            "domain": DOMAIN_MAP.get(vid, str(vid)),
            "vignette_class": vig_class.get(vid, ""),
        }
        vdf = mdf[mdf["vignette_id"] == vid]
        for prompt in PROMPT_ORDER:
            prow = vdf[vdf["prompt"] == prompt]
            if prow.empty:
                row[f"{prompt}_no_photo"] = None
                for eff in EFFECTS:
                    row[f"{prompt}_{eff}"] = None
                    row[f"{prompt}_{eff}_se"] = None
            else:
                r = prow.iloc[0]
                row[f"{prompt}_no_photo"] = r["no_photo_admit_rate"]
                for eff in EFFECTS:
                    row[f"{prompt}_{eff}"] = r[eff]
                    row[f"{prompt}_{eff}_se"] = r[f"{eff}_se"]
        records.append(row)

    out_df = pd.DataFrame(records)

    out_csv = os.path.join(OUTPUT_DIR, f"vignette_effects_{model_name}.csv")
    out_df.to_csv(out_csv, index=False, float_format="%.4f")

    # ── Pretty-print ──────────────────────────────────────────────────────────
    W = 7  # column width for each effect value
    GAP = "  "

    # Header row 1: prompt labels spanning 4 effect columns each (no_photo + 3 effects)
    span = W * 4 + 6  # 4 values + 3 separators between them
    h1 = f"{'V':>3}  {'Domain':<12}  {'Class':<10}"
    for prompt in PROMPT_ORDER:
        h1 += GAP + f"{prompt.upper():^{span}}"

    # Header row 2: effect names
    h2 = f"{'':>3}  {'':12}  {'':10}"
    for _ in PROMPT_ORDER:
        for col in ["no_photo"] + EFFECTS:
            h2 += GAP + f"{col:>{W}}"

    sep = "─" * len(h1)

    print(f"\n{'═' * len(h1)}")
    print(f"  Model: {model_name}")
    print(f"{'═' * len(h1)}")
    print(h1)
    print(h2)
    print(sep)

    for _, row in out_df.iterrows():
        line = f"{str(row['vignette_id']):>3}  {str(row['domain']):<12}  {str(row['vignette_class']):<10}"
        for prompt in PROMPT_ORDER:
            for col in ["no_photo"] + EFFECTS:
                key = f"{prompt}_{col}"
                val = row.get(key)
                if val is None or (isinstance(val, float) and np.isnan(val)):
                    cell = f"{'—':>{W}}"
                else:
                    fmt = f"{float(val):>{W}.3f}" if col == "no_photo" else f"{float(val):>+{W}.3f}"
                    cell = fmt
                line += GAP + cell
        print(line)

    print(f"\nSaved → {out_csv}")
```

```python
import os

import numpy as np
import pandas as pd

CSV_PATH = "paper_results/med_gemma/Master_Aggregated_Data_Medgemma_EE.csv"
CONTROL_CSV_PATH = "paper_results/med_gemma/Control_Aggregated_Data_EE_medgemma.csv"
OUTPUT_DIR = "."

# Optional: fill in to get human-readable domain labels in the table.
# Keys are vignette_id integers as they appear in the CSV.

DOMAIN_MAP = {
    1:  "TIA",
    2:  "Headache",
    3:  "Appendicitis",
    4:  "DVT",
    5:  "DKA",
    6:  "Pyelonephritis",
    7:  "ITP",
    8:  "Bell's palsy",
    9:  "Uveitis",
    10: "Septic arth.",
    11: "Suicidal id.",
    12: "Psychosis",
}

PROMPT_ORDER = ["baseline", "ignore", "acknowledge"]
EFFECTS = ["IPE", "CE_F", "CE_FvG"]
EPSILON = 1e-5


def _binom_var(p, n):
    return p * (1 - p) / np.maximum(n, 1)


# ── Load data ──────────────────────────────────────────────────────────────────
df_raw = pd.read_csv(CSV_PATH)
ctrl_raw = pd.read_csv(CONTROL_CSV_PATH)

# Control baselines: one row per (model, prompt, vignette_id, condition)
ctrl_pivot = ctrl_raw.pivot_table(
    index=["model", "prompt", "vignette_id"],
    columns="condition",
    values=["admit_rate", "n_total"],
    aggfunc="first",
)
ctrl_pivot.columns = [f"{cond}_{stat}" for stat, cond in ctrl_pivot.columns]
ctrl_pivot = ctrl_pivot.reset_index()

# Vignette class (borderline / clear)
vig_class = (
    df_raw[["vignette_id", "vignette_class"]]
    .drop_duplicates("vignette_id")
    .set_index("vignette_id")["vignette_class"]
)

# ── Per-vignette face mean and sampling variance of that mean ──────────────────
def _face_stats(grp):
    p_i = grp["admit_rate"].values
    n_i = grp["n_total"].values
    k = len(grp)
    p_bar = p_i.mean()
    # Var(p̄) = (1/k²) Σ p_i(1−p_i)/N_i
    var_p_bar = (_binom_var(p_i, n_i)).sum() / (k ** 2)
    return pd.Series({"p_face": p_bar, "var_p_face": var_p_bar, "n_faces": k})


face_df = (
    df_raw.groupby(["model", "prompt", "vignette_id"])
    .apply(_face_stats)
    .reset_index()
)

merged = face_df.merge(ctrl_pivot, on=["model", "prompt", "vignette_id"], how="left")

# Clip baselines to avoid degenerate SE
for cond in ["no_photo", "grey_rect"]:
    merged[f"{cond}_admit_rate"] = merged[f"{cond}_admit_rate"].clip(EPSILON, 1 - EPSILON)

# ── Compute effects ────────────────────────────────────────────────────────────
nop = merged["no_photo_admit_rate"]
gry = merged["grey_rect_admit_rate"]
N_nop = merged["no_photo_n_total"].fillna(1)
N_gry = merged["grey_rect_n_total"].fillna(1)
vpf = merged["var_p_face"]
pf = merged["p_face"]

merged["IPE"]       = (gry - nop).round(4)
merged["IPE_se"]    = np.sqrt(_binom_var(gry, N_gry) + _binom_var(nop, N_nop)).round(4)

merged["CE_F"]      = (pf - nop).round(4)
merged["CE_F_se"]   = np.sqrt(vpf + _binom_var(nop, N_nop)).round(4)

merged["CE_FvG"]    = (pf - gry).round(4)
merged["CE_FvG_se"] = np.sqrt(vpf + _binom_var(gry, N_gry)).round(4)

# ── Build wide output per model ────────────────────────────────────────────────
for model_name in sorted(merged["model"].unique()):
    print(model_name)
    mdf = merged[merged["model"] == model_name]
    vignette_ids = sorted(mdf["vignette_id"].unique())

    records = []
    for vid in vignette_ids:
        row: dict = {
            "vignette_id": vid,
            "domain": DOMAIN_MAP.get(vid, str(vid)),
            "vignette_class": vig_class.get(vid, ""),
        }
        vdf = mdf[mdf["vignette_id"] == vid]
        for prompt in PROMPT_ORDER:
            prow = vdf[vdf["prompt"] == prompt]
            if prow.empty:
                row[f"{prompt}_no_photo"] = None
                for eff in EFFECTS:
                    row[f"{prompt}_{eff}"] = None
                    row[f"{prompt}_{eff}_se"] = None
            else:
                r = prow.iloc[0]
                row[f"{prompt}_no_photo"] = r["no_photo_admit_rate"]
                for eff in EFFECTS:
                    row[f"{prompt}_{eff}"] = r[eff]
                    row[f"{prompt}_{eff}_se"] = r[f"{eff}_se"]
        records.append(row)

    out_df = pd.DataFrame(records)

    out_csv = os.path.join(OUTPUT_DIR, f"vignette_effects_{model_name}.csv")
    out_df.to_csv(out_csv, index=False, float_format="%.4f")

    # ── Pretty-print ──────────────────────────────────────────────────────────
    W = 7  # column width for each effect value
    GAP = "  "

    # Header row 1: prompt labels spanning 4 effect columns each (no_photo + 3 effects)
    span = W * 4 + 6  # 4 values + 3 separators between them
    h1 = f"{'V':>3}  {'Domain':<12}  {'Class':<10}"
    for prompt in PROMPT_ORDER:
        h1 += GAP + f"{prompt.upper():^{span}}"

    # Header row 2: effect names
    h2 = f"{'':>3}  {'':12}  {'':10}"
    for _ in PROMPT_ORDER:
        for col in ["no_photo"] + EFFECTS:
            h2 += GAP + f"{col:>{W}}"

    sep = "─" * len(h1)

    print(f"\n{'═' * len(h1)}")
    print(f"  Model: {model_name}")
    print(f"{'═' * len(h1)}")
    print(h1)
    print(h2)
    print(sep)

    for _, row in out_df.iterrows():
        line = f"{str(row['vignette_id']):>3}  {str(row['domain']):<12}  {str(row['vignette_class']):<10}"
        for prompt in PROMPT_ORDER:
            for col in ["no_photo"] + EFFECTS:
                key = f"{prompt}_{col}"
                val = row.get(key)
                if val is None or (isinstance(val, float) and np.isnan(val)):
                    cell = f"{'—':>{W}}"
                else:
                    fmt = f"{float(val):>{W}.3f}" if col == "no_photo" else f"{float(val):>+{W}.3f}"
                    cell = fmt
                line += GAP + cell
        print(line)

    print(f"\nSaved → {out_csv}")
```

## Per race/gender CE effects

```python
import pandas as pd
import numpy as np

def generate_demographic_ce_table(
    master_csv="paper_results/Master_Aggregated_Data_EE.csv", 
    control_csv="Control_Aggregated_Data_updated.csv"
):
    print(f"Loading data from {master_csv} and {control_csv}...\n")
    try:
        df_raw = pd.read_csv(master_csv)
        ctrl_raw = pd.read_csv(control_csv)
    except FileNotFoundError as e:
        print(f"Error: {e}")
        return

    # 1. Map photo_id to Race and Gender
    race_map = {'A': 'Asian', 'B': 'Black', 'I': 'Indian', 'L': 'Latino', 'W': 'White'}
    gender_map = {'M': 'Male', 'F': 'Female'}
    
    df_raw['Race'] = df_raw['photo_id'].str[0].map(race_map)
    df_raw['Gender'] = df_raw['photo_id'].str[1].map(gender_map)
    
    # Drop rows where photo_id wasn't standard
    df_raw = df_raw.dropna(subset=['Race', 'Gender'])
    
    # Handle column naming dynamically (could be admit_rate or admit)
    admit_col = 'admit_rate' if 'admit_rate' in df_raw.columns else 'admit'
    ctrl_admit_col = 'admit_rate' if 'admit_rate' in ctrl_raw.columns else 'admit'

    # 2. Isolate the no_photo baseline from control data
    ctrl_no_photo = ctrl_raw[ctrl_raw['condition'] == 'no_photo'].copy()
    
    # Average baseline across control runs (just in case there are multiple runs per vignette)
    ctrl_baseline = ctrl_no_photo.groupby(['model', 'prompt', 'vignette_id'])[ctrl_admit_col].mean().reset_index()
    ctrl_baseline.rename(columns={ctrl_admit_col: 'no_photo_admit'}, inplace=True)

    # 3. Merge the baseline admission rates into the master data
    merged = pd.merge(df_raw, ctrl_baseline, on=['model', 'prompt', 'vignette_id'], how='left')

    # 4. Calculate Marginal Causal Effect (CE) per instance
    merged['CE'] = merged[admit_col] - merged['no_photo_admit']

    # 5. Group by Model, Race, Gender, and Prompt
    grouped = merged.groupby(['model', 'Race', 'Gender', 'prompt'])[[admit_col, 'CE']].mean().reset_index()

    # 6. Generate the Markdown Table per Model
    prompts = ['baseline', 'acknowledge', 'ignore']
    
    for model_name in grouped['model'].unique():
        print(f"### Marginal Causal Effect (CE) per Race and Gender - Model: {model_name}")
        
        model_df = grouped[grouped['model'] == model_name]
        
        # Pivot the table so prompts become columns
        pivot = model_df.pivot(index=['Race', 'Gender'], columns='prompt', values=[admit_col, 'CE'])
        
        # Print Markdown Table Header
        print("| Race | Gender | Baseline Admit | Baseline CE (pp) | Ack Admit | Ack CE (pp) | Ignore Admit | Ignore CE (pp) |")
        print("|---|---|---|---|---|---|---|---|")
        
        # Print each row
        for (race, gender), row in pivot.iterrows():
            
            # Helper function to safely pull and format data
            def get_val(metric, p_type, is_ce=False):
                if (metric, p_type) in row and pd.notna(row[(metric, p_type)]):
                    val = row[(metric, p_type)]
                    if is_ce:
                        return f"{val * 100:+.1f}" # Convert to pp and force +/- sign
                    else:
                        return f"{val:.3f}"
                return "—"

            b_adm = get_val(admit_col, 'baseline')
            b_ce  = get_val('CE', 'baseline', is_ce=True)
            
            a_adm = get_val(admit_col, 'acknowledge')
            a_ce  = get_val('CE', 'acknowledge', is_ce=True)
            
            i_adm = get_val(admit_col, 'ignore')
            i_ce  = get_val('CE', 'ignore', is_ce=True)
            
            print(f"| {race} | {gender} | {b_adm} | {b_ce} | {a_adm} | {a_ce} | {i_adm} | {i_ce} |")
        
        print("\n")

if __name__ == "__main__":
    generate_demographic_ce_table()
```
